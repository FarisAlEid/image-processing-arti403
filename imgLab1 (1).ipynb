## Load using website code

```python
import matplotlib.pyplot as plt #for creating and displaying plots
import matplotlib # Import Matplotlib itself to modify global plotting settings
import numpy as np #  for numerical and array operations
import skimage as ski #  scikit-image provides sample images and image-processing tools


matplotlib.rcParams['font.size'] = 18 #  font size 

img = ("camera") # name of the sample image

caller = getattr(ski.data, img)# Get the image loading function from skimage.data using the image name


image = caller() #Load the selected sample image

plt.figure() # Create a new figure for displaying the image

plt.title(img) # Set the title of the figure to the image name


# Check whether the image is grayscale or colored
# A grayscale image has 2 dimensions (height × width)
if image.ndim == 2:
    # Display the grayscale image using a gray colormap
    plt.imshow(image, cmap=plt.cm.gray)

else:
    # Display a color image using its original colors
    plt.imshow(image)


# Display image 
plt.show()
```
